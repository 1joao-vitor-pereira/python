```python
### Part 1 Building the base Anagram Checker

## Anagram checker to ensure words identified as anagrams are indeed anagrams.

# Prompt the user to enter two words

def are_anagrams():
        """"
        Description:
        This checker is dedicated to identify if two words are anagrams of each other. An anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once
    
        Args:   
        word1 (str): The first word to compare.
        word2 (str): The second word to compare.
    
        Returns: 
        bool: True if the words are anagrams, False otherwise.
        """

        # Prompt the user to enter two words
        word1 = input("Enter the first word: ")
        word2 = input("Enter the second word: ")
        
        # Remove spaces and convert to lowercase for accurate comparison
        word1 = word1.lower()
        word2 = word2.lower()
        
        # Sort the characters of both words and compare and print the result
        if sorted(word1) == sorted(word2):
                print(f"{word1} and {word2} are anagrams.")
        else:
                print(f"{word1} and {word2} are not anagrams.")

#Call the are anagrams function and print the result
are_anagrams()
```

```python
### Part 2 Expanding the functionality of the Anagram Checker

## Anagram checker to ensure words identified as anagrams are indeed anagrams and if it's case sensitivbe or not.

#Prompt the user to enter two words and if it's case sensitive or not

def anagram_checker():
    """
    Description:
    This checker is dedicated to identify if two words are anagrams of each other. An anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once.

    Args:   
    word1 (str): The first word to compare.
    word2 (str): The second word to compare.
    is_case_sensitive (bool): If True, case matters. If False, case is ignored.

    Returns: 
    bool: True if the words are anagrams, False otherwise.
    """

    # Prompt the user to enter two words
    word1 = input("Enter the first word: ")
    word2 = input("Enter the second word: ")
    
    # is_case_sensitive = input("Is it case sensitive? (True/False): ") == "True"
    is_case_sensitive = input("Is it case sensitive? (True/False): ") == "True"

    # If case is NOT sensitive, lowercase both words before comparing
    if not is_case_sensitive:
        word1 = word1.lower()
        word2 = word2.lower()

    # Sort the characters of both words and compare and print the result
    if sorted(word1) == sorted(word2):
        print(f"{word1} and {word2} are anagrams.")
    else:
        print(f"{word1} and {word2} are not anagrams.")

#Call the are anagrams function and print the result based on the case sensitiveness
anagram_checker()
```
